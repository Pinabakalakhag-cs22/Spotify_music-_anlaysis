Feature Engineering


```python
df["duration_min"]=(df["duration_ms"] / 60000).round(2)
df[["track_name","duration_ms","duration_min"]].head()
```

```python
df["artist_count"]=df['artists'].str.split(';').str.len()
df[["artists","artist_count"]].head()
```

```python
df["duartion_category"]=pd.cut(
    df["duration_min"],
    bins=[-0.01,3,5,10,np.inf],
    labels=["short","medium","long","very long"]
)
df[["track_name","duration_min","duartion_category"]].head()
```

```python
df["popularity_category"]=pd.cut(
    df["popularity"],
    bins=[-0.01,30,60,100],
    labels=["low","medium","high"]
)
df[["track_name","popularity","popularity_category"]].head()
```

```python
df['energy_category'] = pd.cut(
    df['energy'],
    bins=[-0.01, 0.30, 0.60, 0.80, 1.01],
    labels=['Low', 'Medium', 'High', 'Very High']
)
df[['track_name','energy',
    'energy_category']].head()
```

```python
df['danceability_category'] = pd.cut(
    df['danceability'],
    bins=[-0.01, 0.30, 0.60, 0.80, 1.01],
    labels=['Low', 'Medium', 'High', 'Very High']
)
df[['track_name','danceability',
    'danceability_category']].head()
```

```python
df['mood'] = np.where(
    df['valence'] >= 0.60,
    'Happy',
    np.where(
        df['valence'] <= 0.30,
        'Sad',
        'Neutral'
    )
)

df[['track_name', 'valence', 'mood']].head()
```

```python
df['tempo_category'] = pd.cut(
    df['tempo'],
    bins=[-0.01, 80, 120, 160, np.inf],
    labels=['Slow', 'Moderate', 'Fast', 'Very Fast']
)
df[['track_name', 'tempo', 'tempo_category']].head()
```

Music kep mapping

```python
key_mapping = {
    0: 'C',
    1: 'C#',
    2: 'D',
    3: 'D#',
    4: 'E',
    5: 'F',
    6: 'F#',
    7: 'G',
    8: 'G#',
    9: 'A',
    10: 'A#',
    11: 'B'
}

df['key_name'] = df['key'].map(key_mapping)
```

Major and minor mode


```python
df['mode_name'] = df['mode'].map({
    0: 'Minor',
    1: 'Major'
})
```

Time signature category

```python
df['time_signature_category'] = df['time_signature'].map({
    3: '3/4',
    4: '4/4',
    5: '5/4'
}).fillna('Other/Unknown')
```

Speechiness category

```python
df['speechiness_category'] = pd.cut(
    df['speechiness'],
    bins=[-0.01, 0.33, 0.66, 1.01],
    labels=['Low', 'Medium', 'High']
)
```

Loudness Category

```python
df['loudness_category'] = pd.cut(
    df['loudness'],
    bins=[-np.inf, -10, -5, np.inf],
    labels=['Quiet', 'Moderate', 'Loud']
)
```

Combimned music feature

```python
df['acoustic_energy_score'] = (
    df['acousticness'] * df['energy']
)

df['mood_energy_score'] = (
    df['valence'] * df['energy']
)

df[['acoustic_energy_score',
    'mood_energy_score']].head()
```

Outiler Detection

```python
features = [
    'popularity',
    'duration_min',
    'danceability',
    'energy',
    'tempo'
]

for col in features:

    plt.figure(figsize=(10,4))

    sns.boxplot(x=df[col])

    plt.title(f'Outlier Detection - {col}')

    plt.show()
```

IQR method

```python
Q1 = df['duration_min'].quantile(0.25)

Q3 = df['duration_min'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = df[
    (df['duration_min'] < lower) |
    (df['duration_min'] > upper)
]

print("Lower limit:", lower)
print("Upper limit:", upper)
print("Outliers:", len(outliers))
```
